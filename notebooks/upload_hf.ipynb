```python
%load_ext autoreload
%autoreload 2

import sys
import os
from pathlib import Path
from dotenv import load_dotenv
load_dotenv()

from IPython.display import display

import matplotlib.pyplot as plt
REPO_DIR = os.environ['REPO_DIR']
DATA_DIR = os.environ['DATA_DIR']
sys.path.append(REPO_DIR)
```

```python
from pathlib import Path
from huggingface_hub import HfApi

api = HfApi()

for model_name in ["google_gemma-3-4b-it", "OpenGVLab_InternVL3-14B"]:
    REPO_ID = f"javifer/{model_name}-saes"
    SAES_ROOT = Path(DATA_DIR) / model_name
    FILES_TO_UPLOAD = ["ae.pt", "config.json", "eval_results.json", "latent_activation_frequency.json"]
    
    for sae_dir in sorted(SAES_ROOT.iterdir()):
        if not sae_dir.is_dir():
            continue
        trainer_dir = sae_dir / "trainer_0"
        if not trainer_dir.is_dir():
            continue
        hub_folder = sae_dir.name
        for name in FILES_TO_UPLOAD:
            f = trainer_dir / name
            if not f.is_file():
                continue
            api.upload_file(
                path_or_fileobj=str(f),
                path_in_repo=f"{hub_folder}/{name}",
                repo_id=REPO_ID,
                repo_type="model",
            )
```
